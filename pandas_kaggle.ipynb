## 2. Indexing, Selecting, and Assigning

Native accessors:   
- [object].column_name
- object['column_name']
- object['column_name'][index]

Indexing in pandas:
- object.iloc[index] --> row in this index
- object.iloc[:, index] --> column i this index (all rows) 

Label-based selection:
- object.loc[index of row, 'column_name'] --> one entry

**Choosing between loc and iloc**
- iloc: Python indexing --> exclusive
- loc: inclusive (can index ANY type, not just numbers but also strings)

**Manipulating and selecting**
